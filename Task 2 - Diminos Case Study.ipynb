```python
data['order_placed_at'] = pd.to_datetime(data['order_placed_at'])
data['order_delivered_at'] = pd.to_datetime(data['order_delivered_at'])
data['delivery_time_mins'] = (data['order_delivered_at'] - data['order_placed_at']).dt.total_seconds() / 60
```

```python
data.info()
```

```python
data
```

```python
percentile_95 = data['delivery_time_mins'].quantile(0.95)
percentile_95
```

```python
data['delivery_time_mins'].describe()
```

```python
late_orders = data[data['delivery_time_mins'] > 31]
late_orders
```

```python
late_orders['delivery_time_mins'].describe()
```

```python
late_percentage = len(late_orders) / len(data) * 100
late_percentage
```

```python
def runningSum(nums):
        result = []
        total = 0        
        for i in nums:
            total += i
            result.append(total)
        return result
```

```python
nums=[1,2,3,4]
runningSum(nums)
```

```python
nums=[1,1,1,1]
runningSum(nums)
```

```python
def shuffle(nums, n):
    result = []
    for i in range(n):
        result.append(nums[i])
        result.append(nums[i + n])
    return result
```

```python
nums = [2, 5, 1, 3, 4, 7]
n = 3
print(shuffle(nums, n))
```

```python
nums = [1,2,3,4,4,3,2,1]
n = 4
print(shuffle(nums, n))
```

```python
def kid(c, ec):
    max_c = max(c)
    result = []
    for i in c:
        if i + ec >= max_c:
            result.append(True)
        else:
            result.append(False)
    return result
```

```python
c = [2, 3, 5, 1, 3]
ec = 3
kid(c, ec)
```

```python
c= [4, 2, 1, 1, 2]
ec = 1
kid(c, ec)
```

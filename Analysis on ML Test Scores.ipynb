```python
data.columns = data.columns.str.strip()
data.Batch.value_counts()
```

```python
# Extract numeric score
data['Score_Num'] = data['Score'].str.split('/').str[0].astype(int)
```

```python
data['Score_Num'].value_counts().sort_index()
```

```python
data.groupby('Batch')['Score_Num'].agg(
    avg_score='mean',
    max_score='max',
    min_score='min',
    students='count'
).sort_values(by='avg_score', ascending=False)
```

```python
def performance(score):
    if score >= 6:
        return 'Excellent'
    elif score >= 4:
        return 'Average'
    else:
        return 'Needs Improvement'

data['Performance'] = data['Score_Num'].apply(performance)
data['Performance'].value_counts()
```
